# Exercise 01:
We have two numbers call a and b. If "a" is even than the number will do multiplication. Else if "a" is odd then the number will do subtraction.

````python
class checking:
  def multiplication(self, a, b):
    return a * b
  def substraction(self, a, b):
    return a - b
  def evenOrOdd(self, a, b):
    if a % 2 == 0:
      result = self.multiplication(a, b)
    else:
      result = self.substraction(a, b)
    return result

c1 = checking()
print("Lets start....")

while True:
  num1 = int(input("Enter the first number: "))
  num2 = int(input("Enter the second number: "))

  if num1 == 0 and num2 == 0:
    print("Please enter atleast one number which is greater then 0.")

  else:
    print(c1.evenOrOdd(num1, num2))
````

# Inheritance

````python
# single class inheritance
class Car:
  def info(self, model, year):
    self.model = model
    self.year = year
    return "Please enter your car info..."
class des(Car):
  def engin_model(self, num):
    self.num = num
    return self.num
c1  = des()
c1.engin_model("Toyota")
c1.info("Corolla", 1097)

print(c1.engin_model("Toyota"))
print(c1.info("Corolla", 1097))
````

````python
# multiple class inheritance
class Car:
  def info(self, model, year):
    self.model = model
    self.year = year
    return "Please enter your car info..."
class des(Car):
  def engin_model(self, num):
    self.num = num
    return self.num
class chasis(des):
  def chasis_number(self, val):
    self.val = val
    return self.val
c1  = chasis()
c1.chasis_number("ABCD")
c1.info("Corolla", 1097)

print(c1.engin_model("Toyota"))
print(c1.info("Corolla", 1097))
print(c1.chasis_number("ABCD"))
````

## Polymorphysm

#Method Overloading
Method overloading allows a class to have multiple methods with the same name, but with different parameters. The compiler or interpreter differentiates between these methods based on the number, type, or order of the arguments passed to them. This is typically resolved at compile time (static polymorphism).

Example: Imagine a class Calculator that can add numbers. You could have:

```
add(int a, int b)
add(int a, int b, int c)
add(double a, double b)
```


When you call add(5, 10), the system knows to use the first method. If you call add(5.0, 10.0), it uses the third.

# Method Overriding
Method overriding occurs when a subclass provides a specific implementation for a method that is already defined in its superclass. The method in the subclass must have the same name, same parameters (signature), and same return type (or a covariant return type) as the method in the superclass. This allows a subclass to provide its own version of an inherited method. This is typically resolved at runtime (dynamic polymorphism).

Example: Consider a superclass Animal with a method makeSound().

```
The Animal class might have a generic makeSound() method.
A Dog subclass could override makeSound() to print("Woof!").
A Cat subclass could override makeSound() to print("Meow!").
```
When you call makeSound() on an object of type Dog, it will execute the Dog's version of the method, not the Animal's.

Key Differences:
* Polymorphism Type: Overloading is static (compile-time), Overriding is dynamic (runtime).
* Signature: Overloaded methods have different signatures (parameters). Overridden methods have the same signature.
* Scope: Overloading occurs within a single class or related classes. Overriding occurs between a superclass and its subclass.

````python
# polymorphysm
# method overwriting
class animal:
  def dog(self):
    return "dog barks too much at mid night."
class animal1(animal):
  def dog(self):
    return "jokhon lej narabey tokhon khabar chachey.", super().dog()
a1 = animal1()
a1.dog() # here we will see "jokhon lej narabey tokhon khabar chachey."
````
