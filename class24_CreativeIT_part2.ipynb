# Importing Libraries

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder

# CROSS-VALIDATION library(train test Split library)
from sklearn.model_selection import train_test_split

from sklearn.linear_model import LinearRegression

from sklearn.metrics import r2_score,mean_squared_error, mean_absolute_error

import pickle

# removing the warning
import warnings
warnings.filterwarnings('ignore')
```

# Initializing Label Encoder

```python
le = LabelEncoder()
```

# Reading Dataset

```python
df = pd.read_csv("Bus Fare.csv")
df.head()
```

Making copy to work on the copied dataset.

```python
df1 = df.copy()
```

# Initial Analysis

```python
df1.isna().sum()
```

```python
df1.duplicated().sum()
```

```python
df1.info()
```

# Encoding a column

```python
df1['Bus_Name_Encoded'] = le.fit_transform(df1['Bus Name'])
```

```python
df1.head()
```

# Renaming The columns

```python
df1 = df1.rename(columns={"Bus Name":"Bus_Name", "Passenger Age":"Passenger_Age", "Distance Travel(km)":"Distance_Travelled", "Bus fare":"Bus_Fare"})
```

```python
df1.head()
```

```python
df1 = df1.rename(columns={"Bus_Name_Encoded":"Bus_Name_Encoded", "Passenger_Age":"Passenger_Age", "Distance_Travelled":"Distance_Travelled", "Bus_Fare":"Bus_Fare"})
```

```python
df1.head()
```

# Separating the dataset

```python
df1.columns
```

```python
x =df1[["Bus_Name_Encoded", " Passanger Age", "Distance_Travelled"]]
y = df1[["Bus_Fare"]]
```

# Split dataset for training-testing

```python
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size = .2, random_state = 42)
```

```python
xtrain.shape, xtest.shape, ytrain.shape, ytest.shape
```

# Model Train

```python
lr = LinearRegression()
```

```python
model = lr.fit(xtrain, ytrain)
```

R2 score for train data

```python
model.score(xtrain, ytrain)
```

Check Test Result

```python
model.score(xtest, ytest)
```

```python
df1.head(3)
```

# Checking Prediction

```python
model.predict([[5, 25, 50]])
```

# Evaluate the Model

```python
y_pred = model.predict(xtest)
y_pred
```
