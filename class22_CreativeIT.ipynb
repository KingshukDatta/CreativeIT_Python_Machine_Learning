autopct="%1.2f%%" -> after point how many digits will show

```python
plt.figure(figsize=(3,3))
df["sex"].value_counts().plot(kind='pie', autopct="%1.2f%%")
```

```python
plt.figure(figsize=(3,3))
df["sex"].value_counts().plot(kind='pie', autopct="%1.2f%%", startangle= 90)
```

```python
plt.figure(figsize=(3,3))
df["sex"].value_counts().plot(kind='pie', autopct="%1.2f%%", startangle= 90, explode=(0.1, 0))
```

```python
plt.figure(figsize=(3,3))
df["class"].value_counts().plot(kind='pie', autopct="%1.2f%%", startangle= 90, explode=(0.1, 0, 0),shadow = True)
```

```python
plt.figure(figsize=(3,3))
plo = sns.countplot(x=df["sex"])

for p in plo.patches:
    height = p.get_height()
    # Added () to get_width
    plo.text(p.get_x() + p.get_width() / 2, height + 0.2, 
             str(int(height)), ha="center", va="bottom", fontsize=7)
```

# Histogram

```python
sns.histplot(df['age'], kde=True, fill= False, labels= df['sex'])
```

```python
sns.FacetGrid(df, hue='survived')\
  .map(sns.histplot, 'age', kde=True, bins = 25,)\
    .add_legend()
```

```python
sns.FacetGrid(df, hue='survived')\
  .map(sns.histplot, 'age', kde=True, bins = 25, fill=False)\
    .add_legend()
```
