```python
# List
a = [1,2, "rudra", False, True]
print(a)
print(type(a))
```

```python
# List Function
s = list([1,2,"Kingshuk", True, False])
print(s)
print(type(s))
```

```python
# List Function
# Nested list

S =list([1,"king", False,[1,"King"]])
print(S)
print(type(S))
```

```python
# index number
# python's index number starts from 0 to infinite --> Going Forward || -1 to minus infinite --> Goingbackward
# nested list can be defined by s[][]...[]

t =list([1,"king", False,[1,"King"]])
print(t[1])
t[3][1]
```
