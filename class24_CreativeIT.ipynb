## Properties of Linear Regression
In the linear regression line, if the regression parameters a₀ and a₁ are defined, the properties are given below:

* The linear regression line reduces the sum of squared differences between observed values and predicted values.
* The linear regression line always passes through the mean of the X and Y variable values.
* The linear regression constant (b₀) is equal to the y-intercept of the linear regression.
* The linear regression coefficient (b₀) is the slope of the regression line.

# Loss/Error/Residual:
loss (or a loss function) is a measure of how far off a model's prediction is from the actual truth. It quantifies the "error" for a single training


## Mean Squared Error (MSE)
This is the most common loss function used for Regression (predicting numerical values like house prices). It squares the difference between the predicted and actual values to ensure the result is always positive and to penalize larger errors more heavily.



Where:
* $n$ = Number of data points.
* $y_i$ = The actual observed value (the ground truth).
* $\hat{y}_i$ = The predicted value from your model.
* $\sum$ = The sum of all errors from $i=1$ to $n$.

## Mean Absolute Error (MAE)
MAE stands for Mean Absolute Error. It is a common metric used in regression analysis to measure the average magnitude of errors in a set of predictions, without considering their direction (positive or negative).

Unlike MSE (Mean Squared Error), which squares the errors, MAE treats all errors linearly. This makes it more robust to outliers because it doesn't penalize large errors as heavily as squaring does.



Where:
* $n$: The total number of observations or data points.
* $\sum$: The sigma symbol, representing the sum of all values from $i=1$ to $n$.
* $y_i$: The actual true value (ground truth).
* $\hat{y}_i$: The predicted value generated by the model.
* $| \cdot |$: The absolute value bars, which turn any negative difference into a positive one.
