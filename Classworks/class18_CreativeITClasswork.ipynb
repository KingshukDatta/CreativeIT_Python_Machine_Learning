# Ordinal Encoding

````python
# using manual approach by replace method.
df1["Stress_Level"]= df1["Stress_Level"].replace(["Low","Medium","High"],[1,2,3])
df1.head()
````

````python
# using loop approach for ordinal encoding

colForEncode = ["Cholesterol_Level","Exercise_Frequency","Sleep_Hours"]

for col in colForEncode:
  df1[col + '_Encoded'] = encoder.fit_transform(df1[[col]])

df1.head()
````

# One Hot Encoding

````python
# using the "df2" copied dataframe for one hot encoding.
df3.head()
````

````python
# Manual one-hot encoding for Cholesterol_Level
df3['Cholesterol_Level_Low'] = (df3['Cholesterol_Level'] == 'Low').astype(int)
df3['Cholesterol_Level_Medium'] = (df3['Cholesterol_Level'] == 'Medium').astype(int)
df3['Cholesterol_Level_High'] = (df3['Cholesterol_Level'] == 'High').astype(int)
df3['Cholesterol_Level_Borderline'] = (df3['Cholesterol_Level'] == 'Borderline').astype(int)

# Drop the original column
df3 = df3.drop('Cholesterol_Level', axis=1)

# View the result
df3.head()
````

````python
# One hot encoding using loop
colForOHE = ["Exercise_Frequency","Sleep_Hours"]

for col in colForOHE:
  # Fit and transform the column using OneHotEncoder
  encoded_data = ohe.fit_transform(df3[[col]])
  
  # Get the feature names for the encoded columns
  feature_names = ohe.get_feature_names_out([col])
  
  # Create a DataFrame for the encoded data with the same index as df3
  encoded_df = pd.DataFrame(encoded_data, columns=feature_names, index=df3.index)
  df3 = pd.concat([df3, encoded_df], axis=1)
  
# Drop the original columns
df3 = df3.drop(colForOHE, axis=1)

df3.head()
````

# Min Max scaler

````python
# manual approach for min-max scaling
# electing column that are needed to be scaled
col_name = 'Blood_Sugar_Level'

# getting min and max value of the column
min_value = df4[col_name].min()
max_value = df4[col_name].max()
````

* The formula of Min Max Scaling: 

$$X_{new} = \frac{X - X_{min}}{X_{max} - X_{min}}$$

````python
# Applying the min-max scaling formula to the column
df4[col_name + '_Scaled'] = (df4[col_name] - min_value) / (max_value - min_value)
df4.head()
````

````python
# checking the results of scaling
print(df4[[col_name, col_name + '_Scaled']].head())
print("Min:", df4[col_name + '_Scaled'].min())
print("Max:", df4[col_name + '_Scaled'].max())
````
