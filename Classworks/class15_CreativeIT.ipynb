Using loop

```python
for col in df2.columns:
  if is_numeric_dtype(df2[col]):
    continue
  else:
    count = df2[col].nunique()
    cate_list = df2[col].unique().tolist()
    if count == 2:
      df2[col] = df2[col].replace(cate_list, [0, 1])
    elif count == 3:
      df2[col] = df2[col].replace(cate_list, [0, 1, 2])
    elif count == 4:
      df2[col] = df2[col].replace(cate_list, [0, 1, 2, 3])
```
