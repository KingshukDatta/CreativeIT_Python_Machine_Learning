## Grading System
x < 40 -> Fail
x: 40-50 -> B+
51-59 -> A-
60-69-> A
70-79 -> A+
x>= 80 -> Outstanding

````python
x = float(input("Enter a grade: "))

# Condition checking

if x < 40:
  print("Fail.")
elif (x >=40 and x <=50):
  print("B+")
elif (x >=51 and x <=59):
  print("A-")
elif(x >=60 and x <=69):
  print("A")
elif(x >=70 and x <=79):
  print("A+")
else:
  print("Outstanding")
````

# Weather forcasting

### Question: Weather Forecast

Write a Python program that takes the current temperature as input and prints a weather-related suggestion based on the following conditions:

*   If the temperature is below 0°C, suggest "Wear heavy winter clothes."
*   If the temperature is between 0°C and 10°C (inclusive), suggest "Wear a warm jacket."
*   If the temperature is between 11°C and 20°C (inclusive), suggest "A light jacket will be fine."
*   If the temperature is above 20°C, suggest "Enjoy the warm weather!"

Example Input:
```
15
```

Expected Output:
```
A light jacket will be fine.
```

````python
temp = float(input("Enter the temperature: "))

# Condition starts
if temp < 0:
  print('Wear heavy winter clothes.')
elif(temp>=0 and temp<=10):
  print('Wear a warm jacket.')
elif(temp>=11 and temp<=20):
  print('A light jacket will be fine.')
else:
  print('Enjoy the warm')
````

````python
b = "I love"

for char in b[::-1]:
  print(char)
````

````python
# Continue
for i in range(1,50):
  if i == 20 or i == 25:
    continue # 20 is skipped
  print(i)
````

````python
# Break
for i in range(1,50):
  if i == 20:
    break # 20 is skipped
  print(i)
````
