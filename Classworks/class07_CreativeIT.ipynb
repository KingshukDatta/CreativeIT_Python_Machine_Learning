# Non-Parameterized Function
Here we can not edit or give a value to a function.

````python
def addi():
  a = 10
  b = 15
  c = a + b
  return c
addi()
````

````python
def addiction():
  a = [1, 2, 3, 4, 5]
  total = 0
  for i in a:
    total += i
  return total


addiction()
````

````python
def prime(n):
    if n <= 1:
        print("Not Prime")
        return

    for i in range(2, n):
        if n % i == 0:
            print("Not Prime", n)
            break
    else:
        print("Prime", n)

prime(10)
````

````python
for i in range(11,20):
  for j in range(2,i):
    if i % j == 0:
      print("not prime", i)
      break
  else:
    print("prime",i)
````

# Parameterized Function
Here we can give value or parameter to a function.

````python
def name(a, b):
  return a, b

name("Kingshuk", "Supti")
````

````python
name(25,26)
````

````python
def addiction(a):
  total = 0
  for i in a:
    total += i
  return total

a = [1, 2, 3, 4, 5]
addiction(a)
````

# Argument Type:


1.   Positional Argument.
2.   Keyword Argument.
3.   Variable Length Argument.
4.   Variable Length Keyword Argument.



# Positional Argument:
When you call a function with arguments without using keywords, they are called positional arguments.Positional arguments must be in the correct order.

````python
# Positional Argument.

def info(name, age):
  return name, age

info("Kingshuk", 25)
````

# Keyword Argument.
When you call a function with arguments with using keywords, they are called keyword arguments.With keyword arguments, the order of the arguments does not matter.

````python
# Keyword Argument.

def info(name, age):
  return name, age

info(age = 25, name = "Kingshuk")
````
