```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
df= sns.load_dataset('titanic')
```

```python
df.head()
```

# kde plot

```python
sns.kdeplot(x= 'age', y= 'fare', data = df)
```

```python
sns.kdeplot(x= 'age', y= 'fare', data = df, fill= True)
```

```python
sns.kdeplot(x= 'age', y= 'fare', data = df, fill= True, hue = "sex")
```
