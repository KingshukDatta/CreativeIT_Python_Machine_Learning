# Visualization

```python
# line plot
plt.plot(df["age"])
```

```python
x = [10, 15, 20, 25] # x-axis values: price
y = [2, 3, 4, 5] # y-axis values: quantity
# line plot: plt.plot(x, y)
# scatter plot
plt.figure(figsize=(10, 5)) # to increase the size of the plot. (row, column)
plt.plot(x, y, "o", color='r', marker="*",)

# to add title and labels to the plot
plt.title("Price vs Quantity")
plt.xlabel("value of the product")
plt.ylabel("quantity of the product")

# not showing unwanted writing like: Text(0, 0.5, 'Quantity')
plt.show()
```

```python
sns.lineplot(x=x, y=y)
```

```python
plt.plot(df["age"], df["bmi"])
```

```python
plt.plot(df["age"], df["bmi"], "o")
```

```python
sns.lineplot(x=df["age"], y=df["bmi"])
```

```python
# scatter plot
plt.scatter(df["age"], df["bmi"], color='green', alpha=0.5) # alpha is used to set the transparency of the points. (0 to 1)
```

```python
# same thing for seaborn
plt.figure(figsize=(10, 5))
sns.scatterplot(x=df["age"], y=df["bmi"], color='green', alpha=0.5, size=df["children"], hue=df["sex"]) # size is used to set the size of the points. (default is 20)
plt.xlabel("Age")
plt.ylabel("BMI")
plt.title("Age vs BMI")
plt.legend(loc='upper left') # to set the position of the legend. (upper left, upper right, lower left, lower right, center)
plt.show()
```

```python
sns.scatterplot(x=df["age"], y=df["bmi"], color='green', alpha=0.5, size=df["children"], hue=df["sex"], style=df["smoker"], palette="Set1") # style is used to set the style of the points. (default is "o")
plt.xlim([10, 30])
plt.ylim([10, 30])
```

```python
# bar plot
x = ["Apple", "Banana", "Orange", "Grapes"]
y = [5, 20, 15, 20]
plt.bar(x, height=y, bottom=10)
```

```python
# bar plot
x = ["Apple", "Banana", "Orange", "Grapes"]
y = [5, 20, 15, 20]
plt.bar(x, y, color='r')
```

```python
# bar plot
x = ["Apple", "Banana", "Orange", "Grapes"]
y = [5, 20, 15, 20]
# to set different colors for each bar, we can use a list of colors.
colors = ['red', 'yellow', 'orange', 'purple']
plt.bar(x, y, color=colors, width=0.5) # width is used to set the width of the bars. (default is 0.8)
```
