# "enumerate()"-functuon
The Python enumerate() function is a built-in utility that adds a counter to an iterable (like a list, tuple, or string) and returns an enumerate object containing pairs of an index and the corresponding value. This is a "Pythonic" way to iterate through a sequence while keeping track of both the position and value simultaneously, simplifying code and avoiding manual index management.

fruits = ['apple', 'banana', 'cherry']
for index, fruit in enumerate(fruits):
    print(f"Index {index}: {fruit}")

## find the vowel

````python
b  = 'I love machine Learning'
v = ["a","e","i", "o", "u", "A", "E", "I", "O", "U"]
for i in b:
  if i in v:
    print(f"The character '{i}' is a vowel")

  elif i == " ":
    continue

  else:
    print(f"The character '{i}' is a consonant")
````
