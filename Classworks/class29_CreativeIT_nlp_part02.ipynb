# Sentence Creation
A sentence has been created to go further.

```python
a = "I love Machine learning. It is amazing." 
```

# Tokenization

## Word Tokenization
Word token is used to extract words from a  token.

```python
word_token = word_tokenize(a)
word_token
```

## Sentence Tokenization
Sentence token is used to convert word tokens into sentence.

```python
sen_token = sent_tokenize(a)
for index, sentence in enumerate(sen_token):
    print(f"Sentence {index + 1}: {sentence}")
```

```python
sen_token = sent_tokenize(a)
for index, sentence in enumerate(sen_token):
  print(f"Sentence {index + 1}: {sentence}")
```

## Creating list and storing word tokens
creating a list called "li" and storing all the word token. we used .lower() to turn every word to lower case.

```python
li= []
word_token = word_tokenize(a)
for token in word_token:
    li.append(token.lower())
li
```

# Data cleaning
Data is being cleaned by removing punctuations and unnecessary items.

```python
### Gemini ###
# 1. Create an empty list to store the cleaned words
cleaned_tokens = []

for token in word_token:
    # Remove punctuation
    clean_text = re.sub(r'[^\w\s]', '', token)
    
    # 2. Only append if the token isn't completely empty (e.g., punctuation marks become '')
    if clean_text: 
        cleaned_tokens.append(clean_text)

# 3. Join the entire list of cleaned words back into a sentence
clen_data = " ".join(cleaned_tokens)

clen_data
```
