# Exception Handling
for exception handling we use "try" and "except". It is just like "if-else" condition.

```python
# Global variables
a = 10
b = 15

def add(c,d):
    print(c + d)
    # f string is used for handling both string and variable together
    print(f"Sum of {c} and {d} is {c + d}")
    # usig formate() method
    print("Sum of {} and {} is {}".format(c, d, c + d))

add(a,b)
```

```python
a = 10
b = input("Enter any key from your keyboard: ")

try:
  if a%2 ==0:
    c = a/b
    print(c)
except TypeError:
  print("Please enter a number")
```

```python
a = 10
b = input("Enter any key from your keyboard: ")

try:
  if a%2 ==0:
    c = a/b
    print(c)
except Exception as error:
  print("Please enter a number", error)
```

```python
class info:
  def intro(self, name, age, pro):
    self.name = name
    self.age = age
    self.pro = pro
  # for display output
  def display(self):
    return self.name, self.age

# Create an instance of the 'info' class outside its definition
info1 = info()
# Call the 'intro' method on the instance to set its attributes
info1.intro("jack", 100, "bodybuilder")
# Call the 'display' method and print its return value
print(info1.display())
```
