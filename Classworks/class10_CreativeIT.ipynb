# Private attribute and method


```python
class car:
  def __init__(self, name, age):
    self.name = name
    self.age = age
  def show(self):
    print(self.name, self.age)

c1 = car("python", 1991)
c1.show()
```

```python
class car:
  def __init__(self, name, age):
    self.name = name
    self.__age = age


c1 = car("python", 1991)
c1.__age
```

```python
class Outer:
  def __init__(self):
    self.name = "Python"

  class Inner:
    def __init__(self):
      self.name = "guido"

    def show(self):
      print("Hello Welcome to Python!!")

o1 = Outer()
o2 = o1.Inner()
o2.show()
```

```python
class Car:
  def __init__(self, model, price):
    self.model = model
    self.price = price

  def show(self):
    return self.model, self.price
p1 = Car("BMW", 1000000)
print(p1.show())
```
