```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.tag import pos

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
a = "I love Machine learning. It is amazing." 
```

```python
word_token = word_tokenize(a)
word_token
```

```python
sen_token = sent_tokenize(a)
for index, sentence in enumerate(sen_token):
    print(f"Sentence {index + 1}: {sentence}")
```

```python
sen_token = sent_tokenize(a)
for index, sentence in enumerate(sen_token):
  print()
```

```python
li= []
word_token = word_tokenize(a)
for token in word_token:
    li.append(token.lower())
li
```

```python
import re
for token in word_token:
    clean_text = re.sub(r'[^\w\s]', '', token)
    print(clean_text)
clen_data = " ".join(clean_text)

clen_data
```
