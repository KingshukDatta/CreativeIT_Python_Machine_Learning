# Question 01: Check Even or Odd
Write a program to check if a given number is even or odd.

```python
# Declaring Variables and Taking Input.
num = float(input("Enter a number: "))

# Checking if the number is even or odd.
if num %2 == 0:
  print("The Number ", num, " is Even.")
else:
  print("The Number ", num, " is Odd.")
```

# Question 02: Positive, Negative, or Zero
Determine if a number is positive, negative, or zero.

```python
# Declaring Variables and Taking Input.
value = float(input("Enter a number: "))

# Checking if the number is positive, negative or zero.
if value > 0:
  print("The Number ", value, " is Positive.")
elif value < 0:
  print("The Number ", value, " is Negative.")
else:
  print("The Number ", value, " is Zero.")
```

# Question 03: Vowel or Consonant
Write a program to check whether a character is a vowel or consonant.

```python
# Declaring List
vowel = ['a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U']
# Declaring Variables and Taking Input.
while True:
    word = input("Enter exactly one character: ")
    if len(word) == 1:
        char = word
        break
    else:
        print("Invalid input. Please try again.")

# Checking if the word is vowel or consonent.
if char in vowel:
    print(char, "is a vowel.")
else:
    print(char, "is a consonent.")
```

# Question 04: Leap Year
Check if a given year is a leap year or not.

```python
from ast import And
# Declaring Variables
year = int(input("Enter a year: "))

# Checking if the year is leap year or not.
if (year % 4 == 0) and (year % 100 != 0 or year % 400 == 0):
  print(year, "is a leap year.")
else:
  print(year, "is not a leap year.")
```

# Question 05: Grade Evaluator
 Given a score, print the grade (A, B, C, D, or F) based on predefined ranges.

```python
# declaring variables
x = float(input("Enter a grade: "))

# Condition checking

if x < 40:
  print("F")
elif (x >=40 and x <=50):
  print("D")
elif (x >=51 and x <=59):
  print("C")
elif(x >=60 and x <=69):
  print("B")
else:
  print("A")
```

# Question 07: Max of Three Numbers
Find the largest of three numbers.

```python
# declaring variables
num1 = float(input("Enter 1st number: "))
num2 = float(input("Enter 2nd number: "))
num3 = float(input("Enter 3rd number: "))

# Ckecking the largest number.
if (num1 >= num2) and (num1 >= num3):
  largest = num1
elif (num2 >= num1) and (num2 >= num3):
  largest = num2
else:
  largest = num3
print("The largest number is", largest)
```
