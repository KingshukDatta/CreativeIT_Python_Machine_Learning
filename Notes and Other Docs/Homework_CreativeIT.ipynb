```
Golden Rules: As you go down the rows (Row 1, Row 2, Row 3...), the number of spaces and stars changes based on the Row Number.
```

Pattern 01

```
     *
    * *
   * * *
  * * * *
 * * * * *
* * * * * *
```

````python
# Declaring row number.
rows = 6

print("Pattern 01\n")

# loop starts
for i in range(1, rows + 1):
    # (Total rows - current row) -> Total number of spaces needed
    # prints from left to right so
    print(" " * (rows - i), end="")

    # printing star with a space to the end for visual.
    print("* " * i)
````

Pattern 02

```
*********
 *******
  *****
   ***
    *
   ***
  *****
 *******
*********
```

````python
# Declaring Variables
rows = 5

print("--- Pattern 02 ---\n")
# Upper half (Inverted Triangle)
for i in range(rows, 0, -1):
    # Print leading spaces
    print(" " * (rows - i), end="")
    # Print stars (odd numbers: 9, 7, 5...)
    print("*" * (2 * i - 1))

# Lower half (Regular Triangle)
# Start from 2 to avoid duplicating the middle row with 1 star
for i in range(2, rows + 1):
    print(" " * (rows - i), end="")
    print("*" * (2 * i - 1))
````

Pattern 03

```
      *
     * *
    * * *
   * * * *
  * * * * *
 * * * * * *
* * * * * * *
 * * * * * *
  * * * * *
   * * * *
    * * *
     * *
      *
```

````python
rows = 7  # Size of half the diamond

print("--- Pattern 03 ---\n")
# Top half (triangle)
for i in range(1, rows + 1):
    print(" " * (rows - i), end="")
    print("* " * i) # Using "* " creates the spaced diamond look from the image

# Bottom half (Reverse)
for i in range(rows - 1, 0, -1):
    print(" " * (rows - i), end="")
    print("* " * i)
````

Pattern 04

```
            *
          * *
        * * *
      * * * *
    * * * * *
  * * * * * *
* * * * * * *
```

````python
rows = 7

print("--- Pattern 04 ---\n")
for i in range(1, rows + 1):
    # Calculate spaces: Total rows minus current row number
    spaces = rows - i

    # Print spaces first, then stars
    print("  " * spaces + "* " * i)
````

Write a function and return True of if all the items from the given list are prime otherwise print false.
```
a =[10,5,7,4,17]
```
```
b = [11,13,17,19]
```

````python
def is_prime(n):

    if n <= 1:
        return False
    # Check if number is divisible by any number from 2 up to n-1
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

def check_if_all_prime(numbers):

    for num in numbers:
        # If we find even ONE number that is NOT prime, return False immediately
        if not is_prime(num):
            return False

    # If we finish the loop without finding any non-primes, return True
    return True

# Given List
a = [10, 5, 7, 4, 17]
b = [11, 13, 17, 19]

print(f"List 'a' result: {check_if_all_prime(a)}")
print(f"List 'b' result: {check_if_all_prime(b)}")
````
