# List Coprehension
List comprehension is a concise way to create lists in Python. It allows you to generate a new list by applying an expression to each item in an existing iterable (like a list, tuple, or range) and optionally filtering items with a condition.

```
[expression for item in iterable if condition]
```
* expression: The value to add to the new list (e.g., ```x```, ```x*2```, ```str(x)```).

* item: The variable representing the current element being processed.

* iterable: The collection you are looping through.

* condition (optional): A filter; the item is added only if this evaluates to ```True```.

1. Traditional ```For``` Loop:
```
numbers = [1, 2, 3, 4, 5]
squares = []

for n in numbers:
    squares.append(n**2)

# Result: [1, 4, 9, 16, 25]
```

2.  List Comprehension:
```
numbers = [1, 2, 3, 4, 5]
squares = [n**2 for n in numbers]

# Result: [1, 4, 9, 16, 25]
```
Advanced Examples
With Filtering (```if```) Create a list of even numbers only.
```
numbers = range(10)
evens = [n for n in numbers if n % 2 == 0]
# Result: [0, 2, 4, 6, 8]
```
With Conditional Logic (```if...else```) Mark numbers as "Even" or "Odd". Note that when using else, the syntax order changes slightly (the logic moves to the expression part).
```
numbers = [1, 2, 3]
labels = ["Even" if n % 2 == 0 else "Odd" for n in numbers]
# Result: ['Odd', 'Even', 'Odd']
```
***When to Use (and When Not To)***
* **Use it for:** Mapping (converting data types), filtering (removing unwanted data), or flattening simple arrays.

* **Avoid it for:** Complex logic. If your comprehension spans multiple lines or requires nested loops that are hard to read, a traditional for loop is often better for maintainability.

# Built-in Method

list(), filter() and map() method

filter(function_name, variable_name) --> selective work (what we want)
map(function_name, variable_name) --> individual works (returns everything)

````python
li = [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]

def sorted_val(li):
  return li % 2 == 0

va1 = list(filter(sorted_val, li))
va1
````

````python
li = [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]

def square_val(li):
  return li **2

va = map(square_val, li)

````

````python
li = [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]

def sorted_val(li):
  return li % 2 == 0

va1 = list(map(sorted_val, li))
va1
````

# Lambda
anonymous function.

````python
a = range(10)

res = list(filter(lambda x: x % 2 == 0, a))
print(res)
````

````python
li = [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]

def sorted_val(li):
  return li % 2 == 0

va1 = list(map(sorted_val, li))
va1

sq_value = list(map(lambda x: x**2, li))
sq_value
````

````python
a = ["a", "b", "c"]
b = []
for i in a:
  b.append(i.upper())
b
````

````python
# list comprehension way

upp = [i.upper() for i in a]
upp
````

````python
# lambda way
upp_lambda = list(map(lambda i: i.upper(), a))
upp_lambda
````

# Flatten
converting various number of dimension into one dimension.

````python
# Flatten
````

````python
lst = [[1,2,3], [4,5], [6,7,8,9]]
def flatten(lst):
  lis = []
  for sublist in lst:
    if isinstance(sublist, list):
      lis.extend(flatten(sublist))
    else:
      lis.append(sublist)
  return lis
flatten(lst)
````
